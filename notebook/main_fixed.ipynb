## Step 0 - Problem Statement and Proposed Solution

Pemantauan kualitas air permukaan secara kontinu saat ini sangat bergantung pada jaringan sensor Internet of Things (IoT). Meskipun mampu menghasilkan data dalam volume besar dengan resolusi waktu yang tinggi, sistem sensor IoT di lingkungan akuatik sangat rentan terhadap kegagalan teknis, seperti biofouling (penumpukan lumut/bakteri pada probe sensor), kehabisan daya, atau kerusakan kalibrasi, yang menghasilkan pembacaan data yang tidak valid. Di sisi lain, perubahan ekstrem pada kualitas air juga dapat disebabkan oleh kejadian nyata, seperti pembuangan limbah ilegal atau badai hujan yang menyebabkan lonjakan kekeruhan (turbidity) secara drastis. Tantangan komputasional utama dalam ranah ini adalah membedakan antara "fluktuasi ekstrem alami" dan "anomali sesungguhnya" (baik akibat kerusakan sensor maupun pencemaran) di tengah triliunan titik data yang tidak memiliki label (unlabeled data).

Metode deteksi ambang batas statis (static thresholding) tradisional terbukti tidak memadai karena parameter kualitas air bersifat sangat dinamis, multivariat, dan memiliki dependensi temporal (deret waktu). Misalnya, suhu 20°C mungkin **normal di musim panas**, tetapi sangat **abnormal di musim dingin** — meskipun secara global kedua nilai tersebut berada dalam rentang "normal". Anomali semacam ini disebut **anomali kontekstual** (contextual anomaly) dan membutuhkan rekayasa fitur (feature engineering) yang mampu mengekstrak konteks historis, sehingga algoritma dapat memahami pola siklus alami sungai sebelum mencoba mencari keanehan di dalamnya.

Untuk mengatasi ketiadaan label anomali pada data historis, pendekatan Unsupervised Machine Learning seperti Local Outlier Factor (LOF), One-Class Support Vector Machine (OCSVM), dan Isolation Forest (iForest) menjadi solusi yang paling optimal. Namun, evaluasi kinerja model unsupervised seringkali bersifat subjektif. Oleh karena itu, simulasi injeksi anomali sintetik (synthetic anomaly injection) yang merepresentasikan **dua jenis anomali yang berbeda** — (1) anomali global/ekstrem seperti kerusakan sensor dan pencemaran berat, serta (2) anomali kontekstual yang hanya terlihat abnormal relatif terhadap musim/baseline — mutlak diperlukan sebagai ground truth objektif untuk mengukur metrik F1-Score model pada saat pengujian.

Lebih lanjut, setiap algoritma unsupervised memiliki kelemahan bawaan (blind spots); algoritma berbasis kepadatan lokal (LOF) mungkin unggul dalam mendeteksi penyimpangan kontekstual, sementara algoritma berbasis batas (OCSVM) dan berbasis isolasi pohon (iForest) memiliki bias induktif yang berbeda. Untuk menekan tingginya angka False Positive yang sering menjangkiti sistem deteksi anomali tunggal, penerapan **Heterogeneous Ensemble berbasis Voting** (OR, Majority, AND) diusulkan sebagai strategi penggabungan keputusan. Pada akhirnya, kontribusi dari seluruh rekayasa komputasional ini divalidasi melalui studi ablasi (ablation study) untuk membuktikan secara empiris **fitur mana yang benar-benar berkontribusi positif** terhadap peningkatan performa model.

## Step 6 - Imputasi per Subset

Imputasi dilakukan dengan kombinasi `bfill` lalu `ffill`, tetapi **dijalankan terpisah pada setiap subset** (train/val/test). Strategi ini dipilih karena:
- Data bersifat deret waktu yang berkorelasi temporal; nilai sensor cenderung berubah perlahan dari hari ke hari.
- Sederhana, deterministik, dan tidak memerlukan model tambahan.
- **Imputasi per subset mencegah kebocoran (leakage)**: nilai dari periode validation/test tidak akan pernah mengisi nilai kosong di periode training, dan sebaliknya.

```python
print("--- Total NaN per Subset SEBELUM Imputasi ---")
for name, d in [('Train', train_df), ('Val', val_df), ('Test', test_df)]:
    print(f"{name:<6}: {int(d[features].isna().sum().sum())} nilai kosong")

# Imputasi terpisah per subset -> tidak ada nilai yang menyeberang batas periode
train_df = train_df.bfill().ffill()
val_df   = val_df.bfill().ffill()
test_df  = test_df.bfill().ffill()

print("\n--- Total NaN per Subset SESUDAH Imputasi ---")
for name, d in [('Train', train_df), ('Val', val_df), ('Test', test_df)]:
    print(f"{name:<6}: {int(d[features].isna().sum().sum())} nilai kosong")
```

## Step 7 - Synthetic Anomaly Injection

Karena dataset tidak memiliki label anomali, ground truth dibuat secara sintetik. Untuk membuat benchmark yang **realistis**, kita menginjeksikan **dua jenis anomali** dengan proporsi yang seimbang (50:50):

### (a) Anomali Global / Ekstrem

Nilai yang sangat menyimpang dari keseluruhan distribusi, mudah dikenali tanpa konteks temporal.

- **Univariat (Sensor Failure)**: `Turbidity` diset ke 3× kuantil-99 (lonjakan ekstrem). Mensimulasikan biofouling atau probe rusak.
- **Multivariat (Pollution Event)**: `Temperature` naik +6°C **dan** `Dissolved Oxygen` turun −3 mg/L secara bersamaan. Mensimulasikan pencemaran termal.

### (b) Anomali Kontekstual

Nilai yang **masuk akal secara global**, tetapi **abnormal relatif terhadap musimnya**.

- **Univariat (Unusual Temperature for Season)**: `Temperature` diset ke `seasonal_mean + 3σ` dari bulan yang sama — nilainya masih dalam rentang historis tapi jauh menyimpang dari bulan itu.
- **Multivariat (Season Swap)**: `Temperature` dan `Dissolved Oxygen` diset ke **nilai rata-rata bulan berlawanan** (musim dingin di musim panas).

Injeksi dilakukan **pada data mentah** (sebelum feature engineering dan scaling) sehingga semua fitur turunan yang dihitung setelahnya akan otomatis merefleksikan anomali secara konsisten. Training diasumsikan mayoritas normal sehingga seluruhnya diberi label 0; anomali hanya diinjeksikan ke validation dan test.

```python
def inject_anomalies(df_raw_in, n_uni, n_multi, seed):
    rng = np.random.default_rng(seed)
    di = df_raw_in.copy()
    di['Label'] = 0

    idx = di.index            # DatetimeIndex (tz-aware) -> dipakai untuk label .loc
    n   = len(di)

    # Pilih POSISI baris (bukan nilai datetime mentah) agar pelabelan via .loc
    # tetap kompatibel dengan index ber-timezone. Dengan seed yang sama, pemilihan
    # ini identik dengan memilih langsung dari array tanggal.
    uni_pos   = rng.choice(n, size=n_uni, replace=False)
    uni_dates = idx[uni_pos]
    half_uni  = n_uni // 2

    # --- UNIVARIAT ---
    # (a) Global: lonjakan Turbidity ekstrem (sensor failure)
    spike_value = di['Turbidity'].quantile(0.99) * 3
    for date in uni_dates[:half_uni]:
        di.loc[date, 'Turbidity'] = spike_value
        di.loc[date, 'Label'] = 1          # GLOBAL

    # (b) Kontekstual: Temperature tinggi untuk musimnya
    for date in uni_dates[half_uni:]:
        month = pd.Timestamp(date).month
        same_month = di[di.index.month == month]['Temperature']
        di.loc[date, 'Temperature'] = same_month.mean() + 3.0 * same_month.std()
        di.loc[date, 'Label'] = 2          # KONTEKSTUAL

    # --- MULTIVARIAT ---
    remaining_pos = np.where(di['Label'].values == 0)[0]
    multi_pos     = rng.choice(remaining_pos, size=n_multi, replace=False)
    multi_dates   = idx[multi_pos]
    half_multi    = n_multi // 2

    # (a) Global: Temperature +6, DO -3 (pencemaran termal)
    for date in multi_dates[:half_multi]:
        di.loc[date, 'Temperature']      += 6.0
        di.loc[date, 'Dissolved Oxygen'] -= 3.0
        di.loc[date, 'Label'] = 1          # GLOBAL

    # (b) Kontekstual: swap ke bulan berlawanan
    for date in multi_dates[half_multi:]:
        month = pd.Timestamp(date).month
        opposite_month = ((month + 5) % 12) + 1
        opp = di[di.index.month == opposite_month]
        di.loc[date, 'Temperature']      = opp['Temperature'].mean()
        di.loc[date, 'Dissolved Oxygen'] = opp['Dissolved Oxygen'].mean()
        di.loc[date, 'Label'] = 2          # KONTEKSTUAL

    return di

# Training diasumsikan mayoritas normal -> seluruhnya diberi label 0
train_df['Label'] = 0

val_inj  = inject_anomalies(val_df,  n_uni=36, n_multi=36, seed=42)
test_inj = inject_anomalies(test_df, n_uni=72, n_multi=72, seed=143)

print("--- Laporan Ground Truth Buatan ---")
print(f"Validation: {(val_inj['Label']==0).sum()} normal, "
      f"{(val_inj['Label']==1).sum()} global, "
      f"{(val_inj['Label']==2).sum()} kontekstual")
print(f"Test      : {(test_inj['Label']==0).sum()} normal, "
      f"{(test_inj['Label']==1).sum()} global, "
      f"{(test_inj['Label']==2).sum()} kontekstual")
```

## Step 8 - Feature Engineering: Investigasi Dua Pendekatan

Pada tahap ini, tujuan utamanya adalah memperkaya data mentah dengan **konteks temporal** sehingga model dapat membedakan nilai yang "normal untuk musim ini" dari nilai yang "aneh untuk musim ini". Kami menginvestigasi **dua kandidat pendekatan feature engineering**, lalu membandingkannya secara empiris pada ablation study (Step 11).

Agar perbandingan **apple-to-apple**, kedua kandidat sengaja dibuat **identik dalam segala hal kecuali satu**: keduanya memakai **baseline kausal 30 hari yang sama** (rolling window 30 hari atas data sebelum hari-X, via `shift(1)`). Satu-satunya perbedaan adalah **representasi** fitur yang diberikan ke model:

### Kandidat A — Causal Rolling Statistics (30 hari)

Model diberi baseline historis dalam bentuk **mentah**: rata-rata dan standar deviasi 30 hari sebelumnya. Model harus menyimpulkan sendiri seberapa jauh nilai hari-X menyimpang dari baseline tersebut.

$$\text{Mean}_{30D}(x_t) = \frac{1}{30}\sum_{i=t-30}^{t-1} x_i \qquad \text{Std}_{30D}(x_t) = \sigma_{[t-30,\,t-1]}$$

### Kandidat B — Causal Z-Score (30 hari)

Model diberi baseline yang sama, tetapi dalam bentuk **ternormalisasi**: penyimpangan hari-X terhadap baseline dinyatakan eksplisit sebagai z-score.

$$\text{ZScore}(x_t) = \frac{x_t - \text{Mean}_{30D}(x_t)}{\text{Std}_{30D}(x_t)}$$

Perhatikan bahwa z-score hanyalah **kombinasi tetap** dari kedua statistik di Kandidat A. Karena window, kausalitas, dan sumber informasinya identik, ablation study di Step 11 secara bersih mengisolasi satu pertanyaan: *apakah normalisasi eksplisit membantu model unsupervised, ataukah model mampu mempelajari relasi tersebut sendiri dari baseline mentah?*

### Selain itu, kedua kandidat sama-sama dilengkapi:

- **Cyclical encoding bulan** (`Month_Sin`, `Month_Cos`) → menjaga sifat siklik bulan (Desember dekat dengan Januari).

### Catatan kausalitas penting

**Tidak ada `bfill()` pada fitur.** Mengisi baris-baris awal dengan nilai fitur dari hari-hari berikutnya berarti menyuntikkan informasi **masa depan** (look-ahead) — bertentangan dengan sifat kausal yang diklaim. Sebagai gantinya dipakai **warm-up buffer**: fitur untuk validation dihitung dengan menempelkan ~35 hari terakhir **training** di depannya, dan fitur untuk test memakai ~35 hari terakhir **validation**. Buffer berasal dari **masa lalu**, sehingga ini sah dan bukan leakage — yang dilarang hanyalah informasi dari masa depan. Untuk training (tidak punya sejarah sebelum 2014), beberapa hari pertama yang belum memiliki baseline **dibuang**, bukan diisi.

Pada Step 11 nanti, ketiga skenario (**Raw Only**, **Raw + Rolling (Kandidat A)**, **Raw + ZScore (Kandidat B)**) akan dibandingkan head-to-head — dengan hyperparameter yang setara — untuk menentukan mana yang benar-benar meningkatkan F1-Score, **sebelum** tuning dilakukan.


```python
def add_features(df_in, base_features):
    """Hitung fitur turunan secara kausal: baseline 30 hari TANPA hari-X (shift(1))."""
    d = df_in.copy()

    # Cyclical encoding bulan (umum untuk kedua kandidat)
    d['Month_Sin'] = np.sin(2 * np.pi * d.index.month / 12)
    d['Month_Cos'] = np.cos(2 * np.pi * d.index.month / 12)

    # Baseline kausal 30 hari yang SAMA untuk kedua kandidat.
    # shift(1) memastikan hari-X tidak ikut dihitung dalam baseline-nya sendiri.
    for col in base_features:
        rolling_mean = d[col].shift(1).rolling(window=30, min_periods=5).mean()
        rolling_std  = d[col].shift(1).rolling(window=30, min_periods=5).std()

        # Kandidat A: causal rolling statistics 30 hari (baseline bentuk mentah)
        d[f'{col}_Mean_30D'] = rolling_mean
        d[f'{col}_Std_30D']  = rolling_std

        # Kandidat B: causal z-score 30 hari (baseline bentuk ternormalisasi)
        d[f'{col}_ZScore'] = (d[col] - rolling_mean) / (rolling_std + 1e-6)

    # PENTING: tidak ada bfill() di sini. bfill akan mengisi baris-baris awal
    # dengan nilai fitur dari MASA DEPAN (look-ahead) dan melanggar sifat kausal.
    return d


def add_features_with_history(df_target, df_history, base_features, n_buffer=35):
    """Hitung fitur df_target memakai warm-up buffer dari AKHIR df_history.

    Buffer berasal dari masa lalu (subset periode sebelumnya), sehingga BUKAN
    leakage — yang dilarang adalah informasi masa depan, bukan masa lalu.
    Tanpa buffer, ~30 hari pertama tiap subset tidak memiliki baseline.
    """
    buffered = pd.concat([df_history.tail(n_buffer), df_target])
    feat = add_features(buffered, base_features)
    return feat.loc[df_target.index]   # buang kembali baris buffer


# Train tidak punya sejarah sebelum 2014 -> hari-hari pertama yang belum punya
# baseline DIBUANG (bukan diisi dengan nilai masa depan via bfill).
train_feat = add_features(train_df, features)
n_before   = len(train_feat)
train_feat = train_feat.dropna()
print(f"Baris awal train tanpa baseline (dibuang): {n_before - len(train_feat)} hari")

# Validation memakai akhir train sebagai warm-up; test memakai akhir validation.
# Buffer test memakai val_inj (ter-injeksi) agar konsisten dengan prinsip Step 7:
# injeksi dilakukan pada data mentah dan semua turunan merefleksikannya.
val_feat  = add_features_with_history(val_inj,  train_df, features)
test_feat = add_features_with_history(test_inj, val_inj,  features)

# Sanity check: tidak boleh ada NaN yang tersisa di fitur mana pun
assert not train_feat.isna().any().any(), "Masih ada NaN di train_feat"
assert not val_feat.isna().any().any(),   "Masih ada NaN di val_feat"
assert not test_feat.isna().any().any(),  "Masih ada NaN di test_feat"

print("--- Jumlah kolom setelah feature engineering ---")
print(f"Total kolom: {train_feat.shape[1]}")
print()
print("--- Contoh kolom yang tersedia ---")
print(list(train_feat.columns))
```

## Step 9 - Standardization

Algoritma berbasis jarak (LOF) dan berbasis margin (OCSVM) sangat sensitif terhadap skala fitur. `RobustScaler` dipilih karena tahan terhadap outlier (menggunakan median dan IQR, bukan mean dan std). Scaler **hanya difit pada data training** untuk menghindari leakage.

Fitur cyclical (`Month_Sin`, `Month_Cos`) dan kolom `Label` tidak diskalakan.

```python
passthrough_cols = ['Month_Sin', 'Month_Cos', 'Label']
cols_to_scale = [c for c in train_feat.columns if c not in passthrough_cols]

scaler = RobustScaler()
scaler.fit(train_feat[cols_to_scale])

def scale_dataframe(d, scaler_obj, scale_cols, pass_cols):
    arr = scaler_obj.transform(d[scale_cols])
    s = pd.DataFrame(arr, columns=scale_cols, index=d.index)
    return pd.concat([s, d[pass_cols]], axis=1)

train_s = scale_dataframe(train_feat, scaler, cols_to_scale, passthrough_cols)
val_s   = scale_dataframe(val_feat,   scaler, cols_to_scale, passthrough_cols)
test_s  = scale_dataframe(test_feat,  scaler, cols_to_scale, passthrough_cols)

print("--- Contoh Data Train Setelah Scaling ---")
train_s[['Temperature','Turbidity','Temperature_Mean_30D','Temperature_ZScore']].head()
```

## Step 10 - Definisi Feature Sets untuk Ablation

Tiga skenario fitur akan dibandingkan head-to-head di Step 11. Tidak ada satu skenario pun yang diistimewakan: pada ablation, ketiganya akan diuji memakai hyperparameter **default yang identik**. Tuning hyperparameter baru dilakukan **setelah** feature set pemenang ditentukan (Step 12).

Kandidat A dan Kandidat B kini **setara dalam window (30 hari), kausalitas (shift(1)), dan sumber informasi (baseline yang sama)** — satu-satunya perbedaan adalah representasi (statistik mentah vs z-score ternormalisasi).

| Skenario | Fitur | Total |
|---|---|---|
| **Raw Only** | Temperature, DO, pH, Turbidity | 4 |
| **Raw + Rolling** (Kandidat A) | Raw + Mean_30D, Std_30D per parameter + Cyclical | 14 |
| **Raw + ZScore** (Kandidat B) | Raw + ZScore per parameter + Cyclical | 10 |

```python
raw_features     = features  # 4 fitur

rolling_features = (
    features
    + [f'{c}_Mean_30D' for c in features]
    + [f'{c}_Std_30D'  for c in features]
    + ['Month_Sin', 'Month_Cos']
)

zscore_features  = (
    features
    + [f'{c}_ZScore' for c in features]
    + ['Month_Sin', 'Month_Cos']
)

feature_sets = {
    'Raw Only':       raw_features,
    'Raw + Rolling':  rolling_features,
    'Raw + ZScore':   zscore_features,
}

for name, fs in feature_sets.items():
    print(f"{name:<16} ({len(fs)} fitur)")
```

## Step 11 - Ablation Study: Pengaruh Feature Engineering

Tujuan ablation study adalah menguji secara kuantitatif hipotesis dari Step 8.A — bahwa **representasi ternormalisasi (Kandidat B, Causal Z-Score) lebih membantu model daripada baseline mentah (Kandidat A, Causal Rolling Statistics)** pada dataset yang mengandung anomali kontekstual. Karena kedua kandidat memakai window, kausalitas, dan baseline yang identik, perbedaan F1 antar keduanya kini dapat diatribusikan **secara bersih** pada perbedaan representasi tersebut.

Agar perbandingan **adil dan bebas dari sirkularitas**, ablation dijalankan menggunakan **hyperparameter default yang IDENTIK** untuk semua skenario fitur. Dengan begitu, perbedaan F1-Score murni berasal dari kualitas fitur, bukan dari tuning yang menguntungkan salah satu skenario. Tuning baru dilakukan di Step 12 setelah pemenang ditentukan.

> **Catatan metodologis penting:** ablation dievaluasi pada **validation set (2021)**, bukan test set. Pemilihan feature set adalah sebuah **keputusan pemodelan** — sama seperti tuning hyperparameter — sehingga harus diambil tanpa melihat test set. Dengan begitu **seluruh keputusan pemodelan (seleksi fitur + tuning) memakai validation**, dan **test set (2022–2023) tetap murni** sebagai holdout yang baru disentuh pertama kali di Step 13.

```python
# Hyperparameter default yang IDENTIK untuk semua skenario -> perbandingan adil.
ABLATION_PARAMS = {
    'lof':     dict(n_neighbors=20, contamination=0.05, novelty=True),
    'ocsvm':   dict(kernel='rbf', nu=0.05, gamma='scale'),
    'iforest': dict(n_estimators=200, contamination=0.05, random_state=RANDOM_STATE),
}

ablation_results = []
# Ablation dievaluasi pada VALIDATION set -> seleksi fitur tidak menyentuh test.
y_val_binary = (val_s['Label'] >= 1).astype(int)

for scenario_name, feat_subset in feature_sets.items():
    lof = LocalOutlierFactor(**ABLATION_PARAMS['lof'])
    lof.fit(train_s[feat_subset].values)
    p_lof = np.array([1 if x == -1 else 0 for x in lof.predict(val_s[feat_subset].values)])

    svm = OneClassSVM(**ABLATION_PARAMS['ocsvm'])
    svm.fit(train_s[feat_subset].values)
    p_svm = np.array([1 if x == -1 else 0 for x in svm.predict(val_s[feat_subset].values)])

    iso = IsolationForest(**ABLATION_PARAMS['iforest'])
    iso.fit(train_s[feat_subset].values)
    p_iso = np.array([1 if x == -1 else 0 for x in iso.predict(val_s[feat_subset].values)])

    ablation_results.append({
        'Skenario':   scenario_name,
        '#Fitur':     len(feat_subset),
        'F1 LOF':     round(f1_score(y_val_binary, p_lof, zero_division=0), 4),
        'F1 OCSVM':   round(f1_score(y_val_binary, p_svm, zero_division=0), 4),
        'F1 iForest': round(f1_score(y_val_binary, p_iso, zero_division=0), 4),
    })

ablation_df = pd.DataFrame(ablation_results)
ablation_df
```

```python
ab_plot = ablation_df.set_index('Skenario')[['F1 LOF', 'F1 OCSVM', 'F1 iForest']]

ax = ab_plot.plot(kind='bar', figsize=(10, 5), edgecolor='black',
                  color=['#1f77b4', '#ff7f0e', '#2ca02c'])
plt.title("Ablation Study: Effect of Feature Engineering on F1-Score",
          fontweight='bold', fontsize=12)
plt.ylabel("F1-Score (Validation Set)")
plt.xlabel("Skenario Fitur")
plt.xticks(rotation=0)
plt.ylim(0, 1.05)
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.legend(loc='lower right')

for container in ax.containers:
    ax.bar_label(container, fmt='%.3f', padding=3, fontsize=9)

plt.tight_layout()
plt.show()
```

```python
# Pilih feature set pemenang berdasarkan rata-rata F1 ketiga model.
ablation_df['Rata-rata F1'] = ablation_df[['F1 LOF', 'F1 OCSVM', 'F1 iForest']].mean(axis=1)
winner_name = ablation_df.loc[ablation_df['Rata-rata F1'].idxmax(), 'Skenario']
winner_features = feature_sets[winner_name]

print(ablation_df[['Skenario', '#Fitur', 'Rata-rata F1']].to_string(index=False))
print(f"\nFeature set pemenang (berdasarkan F1 validation): '{winner_name}' "
      f"({len(winner_features)} fitur) -> dipakai untuk tuning & evaluasi final.")
```

### Step 11.A - Interpretasi Ablation Study

> ⚠️ **Perlu diverifikasi ulang setelah re-run.** Desain Kandidat A telah diubah (dari rolling 7-hari non-kausal + Diff menjadi causal rolling 30-hari tanpa Diff), sehingga angka dan arah temuan bisa berbeda dari versi sebelumnya. Sesuaikan narasi di bawah dengan hasil aktual.

Karena kedua kandidat kini memakai **baseline yang identik** (window 30 hari, kausal), perbandingan ini mengisolasi satu variabel: **representasi fitur**. Ada dua kemungkinan hasil, dan keduanya merupakan temuan yang sah untuk skripsi:

**Jika Kandidat B (Z-Score) unggul:** model unsupervised berbasis jarak/isolasi terbantu oleh **normalisasi eksplisit** — penyimpangan kontekstual yang semula tersebar di relasi antar-kolom (nilai mentah vs μ vs σ) dipadatkan menjadi nilai ekstrem pada satu sumbu, yang lebih mudah dipisahkan oleh LOF/OCSVM/iForest. Kesimpulannya: *bukan sekadar memberi konteks historis, tetapi **cara merepresentasikan** konteks itulah yang menentukan*.

**Jika kedua kandidat setara (atau Kandidat A unggul):** model ternyata mampu mempelajari relasi deviasi dari baseline mentah, dan keunggulan z-score pada desain lama lebih banyak disebabkan oleh faktor yang dulu tidak terkontrol (window 7 vs 30 hari, non-kausal vs kausal, kehadiran Diff). Ini pun temuan yang jujur dan bernilai metodologis.

**Apa pun hasilnya**, dua hal tetap berlaku:

1. Perbandingan terhadap **Raw Only** menunjukkan apakah konteks temporal (dalam bentuk apa pun) memberikan nilai tambah dibanding nilai mentah saja.
2. Karena ablation memakai hyperparameter setara **dan diukur pada validation set**, kesimpulan pemenang tidak bias terhadap tuning maupun terhadap test set. Test set tetap utuh untuk vonis akhir di Step 13.


### Step 11.B - Ablasi per Tipe Anomali

Memecah hasil studi ablasi berdasarkan tipe anomali (global vs kontekstual) untuk membuktikan bahwa Causal Z-Score terutama meningkatkan deteksi anomali kontekstual. Evaluasi tetap pada **validation set** dengan `ABLATION_PARAMS` (identik antar skenario).

```python
y_val_labels = val_s['Label'].values

ablation_by_type = []

for scenario_name, feat_subset in feature_sets.items():
    lof = LocalOutlierFactor(**ABLATION_PARAMS['lof'])
    lof.fit(train_s[feat_subset].values)
    p_lof = np.array([1 if x == -1 else 0 for x in lof.predict(val_s[feat_subset].values)])

    svm = OneClassSVM(**ABLATION_PARAMS['ocsvm'])
    svm.fit(train_s[feat_subset].values)
    p_svm = np.array([1 if x == -1 else 0 for x in svm.predict(val_s[feat_subset].values)])

    iso = IsolationForest(**ABLATION_PARAMS['iforest'])
    iso.fit(train_s[feat_subset].values)
    p_iso = np.array([1 if x == -1 else 0 for x in iso.predict(val_s[feat_subset].values)])

    for label_val, label_name in [(1, 'Global'), (2, 'Kontekstual')]:
        mask = (y_val_labels == label_val) | (y_val_labels == 0)
        y_sub = (y_val_labels[mask] >= 1).astype(int)

        ablation_by_type.append({
            'Skenario': scenario_name,
            'Tipe': label_name,
            'F1 LOF': round(f1_score(y_sub, p_lof[mask], zero_division=0), 3),
            'F1 OCSVM': round(f1_score(y_sub, p_svm[mask], zero_division=0), 3),
            'F1 iForest': round(f1_score(y_sub, p_iso[mask], zero_division=0), 3),
        })

abl_type_df = pd.DataFrame(ablation_by_type)
print("--- Studi Ablasi per Tipe Anomali ---")
print(abl_type_df.to_string(index=False))

fig, axes = plt.subplots(1, 2, figsize=(14, 5), sharey=True)
for ax, tipe in zip(axes, ['Global', 'Kontekstual']):
    subset = abl_type_df[abl_type_df['Tipe'] == tipe]
    x = np.arange(len(subset))
    width = 0.25
    
    # 1. Simpan objek bar ke dalam variabel
    rects1 = ax.bar(x - width, subset['F1 LOF'].values, width, label='LOF', color='#1f77b4', edgecolor='black')
    rects2 = ax.bar(x,         subset['F1 OCSVM'].values, width, label='OCSVM', color='#ff7f0e', edgecolor='black')
    rects3 = ax.bar(x + width, subset['F1 iForest'].values, width, label='iForest', color='#2ca02c', edgecolor='black')
    
    # 2. Tambahkan label angka tepat di atas setiap bar
    ax.bar_label(rects1, padding=3, fmt='%.3f', fontsize=9)
    ax.bar_label(rects2, padding=3, fmt='%.3f', fontsize=9)
    ax.bar_label(rects3, padding=3, fmt='%.3f', fontsize=9)
    
    ax.set_title(f'Ablasi — Anomali {tipe}', fontweight='bold')
    ax.set_xticks(x)
    ax.set_xticklabels(subset['Skenario'].values, rotation=15)
    ax.set_ylabel('F1-Score')
    
    # 3. Batas Y dinaikkan sedikit ke 1.15 agar angka di atas bar tertinggi tidak terpotong
    ax.set_ylim(0, 1.15) 
    
    ax.legend()
    ax.grid(axis='y', linestyle='--', alpha=0.4)

plt.suptitle('Studi Ablasi Feature Engineering per Tipe Anomali', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

## Step 12 - Hyperparameter Tuning pada Validation Set

Setelah feature set pemenang ditentukan di Step 11, tuning dilakukan pada **validation set 2021** dengan metrik utama F1-Score. Tiga model di-tune secara independen menggunakan `winner_features`:
- **LOF**: `n_neighbors` × `contamination`
- **OCSVM**: `nu` × `gamma`
- **Isolation Forest**: `n_estimators` × `contamination`

### Step 12.A - Tuning LOF

```python
tuning_features = winner_features

lof_results = []
best_lof_f1 = 0
best_lof_params = {}

for n in [10, 15, 20, 25, 30, 50]:
    for c in [0.01, 0.03, 0.05, 0.08, 0.1, 0.15]:
        m = LocalOutlierFactor(n_neighbors=n, contamination=c, novelty=True)
        m.fit(train_s[tuning_features].values)
        p = m.predict(val_s[tuning_features].values)
        p = np.array([1 if x == -1 else 0 for x in p])

        y_val = (val_s['Label'] >= 1).astype(int)
        f1 = f1_score(y_val, p, zero_division=0)
        lof_results.append({'n_neighbors': n, 'contamination': c, 'F1-Score': f1})
        if f1 > best_lof_f1:
            best_lof_f1 = f1
            best_lof_params = {'n_neighbors': n, 'contamination': c}

print(f"[LOF TERBAIK] {best_lof_params} | F1 (val) = {best_lof_f1:.4f}")
```

```python
lof_df = pd.DataFrame(lof_results)
lof_pivot = lof_df.pivot(index='n_neighbors', columns='contamination', values='F1-Score')

plt.figure(figsize=(7, 4.5))
sns.heatmap(lof_pivot, annot=True, fmt=".3f", cmap='YlGnBu', cbar_kws={'label': 'F1-Score'})
plt.title("Heatmap F1-Score Tuning LOF (Validation Set)", fontweight='bold')
plt.xlabel("contamination"); plt.ylabel("n_neighbors")
plt.tight_layout(); plt.show()
```

### Step 12.B - Tuning OCSVM

```python
svm_results = []
best_svm_f1 = 0
best_svm_params = {}

for nu in [0.01, 0.03, 0.05, 0.08, 0.1, 0.15]:
    for g in ['scale', 'auto', 0.01, 0.05, 0.1]:
        m = OneClassSVM(kernel='rbf', nu=nu, gamma=g)
        m.fit(train_s[tuning_features].values)
        p = m.predict(val_s[tuning_features].values)
        p = np.array([1 if x == -1 else 0 for x in p])

        y_val = (val_s['Label'] >= 1).astype(int)
        f1 = f1_score(y_val, p, zero_division=0)
        svm_results.append({'nu': nu, 'gamma': str(g), 'F1-Score': f1})
        if f1 > best_svm_f1:
            best_svm_f1 = f1
            best_svm_params = {'nu': nu, 'gamma': g}

print(f"[OCSVM TERBAIK] {best_svm_params} | F1 (val) = {best_svm_f1:.4f}")
```

```python
svm_df = pd.DataFrame(svm_results)
svm_pivot = svm_df.pivot(index='nu', columns='gamma', values='F1-Score')

plt.figure(figsize=(7, 4.5))
sns.heatmap(svm_pivot, annot=True, fmt=".3f", cmap='YlOrRd', cbar_kws={'label': 'F1-Score'})
plt.title("Heatmap F1-Score Tuning OCSVM (Validation Set)", fontweight='bold')
plt.xlabel("gamma"); plt.ylabel("nu")
plt.tight_layout(); plt.show()
```

### Step 12.C - Tuning Isolation Forest

```python
def evaluate_model(model_obj, name, train_df, test_df, feats):
    preds = model_obj.predict(test_df[feats].values)
    preds = np.array([1 if x == -1 else 0 for x in preds])
    y_true = (test_df['Label'].values >= 1).astype(int)

    print(f"--- LAPORAN KLASIFIKASI {name} (TEST) ---")
    print(classification_report(y_true, preds, target_names=['Normal','Anomali'], zero_division=0))

    # --- Anomaly Score Timeline ---
    scores = -1.0 * model_obj.decision_function(test_df[feats].values)
    labels_raw = test_df['Label'].values  # 0=normal, 1=global, 2=kontekstual

    fig, ax = plt.subplots(figsize=(15, 5))

    ax.plot(test_df.index, scores, color='gray', alpha=0.4, linewidth=0.8,
            label='Anomaly Score')

    ax.axhline(y=0, color='red', linestyle='--', linewidth=1.5, alpha=0.7,
               label='Threshold (Score > 0 = Anomali)')

    mask_global = (labels_raw == 1)
    ax.scatter(test_df.index[mask_global], scores[mask_global],
               color='#1f77b4', s=100, marker='^', edgecolors='black',
               linewidths=0.8, zorder=5, label='Ground Truth: Global')

    mask_ctx = (labels_raw == 2)
    ax.scatter(test_df.index[mask_ctx], scores[mask_ctx],
               color='#ff7f0e', s=100, marker='s', edgecolors='black',
               linewidths=0.8, zorder=5, label='Ground Truth: Kontekstual')

    fn_mask = ((labels_raw >= 1) & (preds == 0))
    if fn_mask.sum() > 0:
        ax.scatter(test_df.index[fn_mask], scores[fn_mask],
                   facecolors='none', edgecolors='red', s=200,
                   linewidths=2, zorder=6, label='False Negative (missed)')

    # Adaptive y-scale untuk mengakomodasi skor ekstrem
    score_p95 = np.percentile(scores, 95)
    score_max = scores.max()
    if score_max > 3 * score_p95 and score_max > 5:
        ax.set_yscale('symlog', linthresh=1)
        max_tick = int(np.ceil(score_max / 10) * 10)
        ticks = [-1, 0, 1, 2, 5, 10]
        if max_tick > 10:
            ticks.extend([20, 50] if max_tick > 20 else [20])
        if max_tick > 50:
            ticks.append(100)
        ax.set_yticks(ticks)
        ax.set_yticklabels([str(t) for t in ticks])
        ax.set_ylabel("Anomaly Score (symlog scale)")
    else:
        ax.set_ylabel("Anomaly Score")

    ax.set_title(f"Anomaly Score Timeline - {name} (Test Set)",
                 fontsize=14, fontweight='bold')
    ax.set_xlabel("Datetime")
    ax.legend(loc='upper right', fontsize=9)
    ax.grid(True, linestyle='--', alpha=0.3)
    plt.tight_layout()
    plt.show()

    cm = confusion_matrix(y_true, preds)
    fig, ax = plt.subplots(figsize=(4.5, 4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False,
                xticklabels=['Normal','Anomali'], yticklabels=['Normal','Anomali'], ax=ax)
    ax.set_xlabel("Prediksi"); ax.set_ylabel("Ground Truth")
    ax.set_title(f"Confusion Matrix - {name}", fontweight='bold')
    plt.tight_layout(); plt.show()

    return preds
```

## Step 14 - Heterogeneous Vote-Based Ensemble

Dengan tiga model heterogen (LOF, OCSVM, iForest) yang memiliki bias induktif berbeda, kita gabungkan prediksinya dengan strategi **voting**. Tiga strategi voting dievaluasi:

- **Vote OR** — anomali jika **setidaknya 1** model menyatakan anomali. Cenderung **menaikkan recall**, cocok ketika False Negative berbahaya.
- **Vote Majority** — anomali jika **minimal 2 dari 3** model setuju. Strategi seimbang antara precision dan recall.
- **Vote AND** — anomali jika **ketiga** model setuju. Cenderung **menaikkan precision**, cocok untuk menekan False Positive.

### Step 14.A - Justifikasi Pemilihan Ensemble Voting

Keputusan untuk menggabungkan LOF, OCSVM, dan Isolation Forest ke dalam ensemble voting didasarkan pada dua argumen teoretis:

**Argumen 1: Condorcet's Jury Theorem (Zimek dkk., 2014)**

Zimek, Campello, dan Sander (2014) dalam survey mereka tentang ensemble untuk unsupervised outlier detection mengangkat Condorcet's Jury Theorem sebagai justifikasi formal: jika setiap "juri" (model) memiliki probabilitas benar > 0.5 dan keputusan mereka independen, maka probabilitas keputusan mayoritas yang benar meningkat seiring bertambahnya juri. Ketiga model kita memenuhi syarat pertama (F1 > 0.5 pada validation set), dan memiliki bias induktif yang fundamental berbeda (kepadatan lokal vs batas marginal vs isolasi pohon) yang secara teoritis mendukung asumsi independensi error.

**Argumen 2: Preseden Empiris (Thomas & Judith, 2020)**

Thomas dan Judith (2020) telah membuktikan secara empiris bahwa voting-based ensemble dari OCSVM + Elliptic Envelope + LOF mampu mengungguli detektor individual pada dataset publik. Pendekatan kita mengikuti kerangka mereka dengan substitusi Elliptic Envelope menjadi Isolation Forest.

**Hipotesis yang Diuji:** *heterogeneous ensemble voting akan menghasilkan F1-Score yang lebih tinggi atau setidaknya lebih stabil dibandingkan model tunggal terbaik, karena setiap model menutupi blind spot model lainnya.* Tiga strategi voting (OR, Majority, AND) dipilih untuk mengeksplorasi trade-off precision-recall yang berbeda sesuai kebutuhan aplikasi.

### Step 14.B - Evaluasi Strategi Voting

```python
y_test = (test_s['Label'].values >= 1).astype(int)

vote_or  = ((pred_lof_test + pred_svm_test + pred_iso_test) >= 1).astype(int)
vote_maj = ((pred_lof_test + pred_svm_test + pred_iso_test) >= 2).astype(int)
vote_and = ((pred_lof_test + pred_svm_test + pred_iso_test) >= 3).astype(int)

print("--- LAPORAN KLASIFIKASI ENSEMBLE VOTE OR ---")
print(classification_report(y_test, vote_or,  target_names=['Normal','Anomali'], digits=3, zero_division=0))
print("--- LAPORAN KLASIFIKASI ENSEMBLE VOTE MAJORITY ---")
print(classification_report(y_test, vote_maj, target_names=['Normal','Anomali'], digits=3, zero_division=0))
print("--- LAPORAN KLASIFIKASI ENSEMBLE VOTE AND ---")
print(classification_report(y_test, vote_and, target_names=['Normal','Anomali'], digits=3, zero_division=0))
```

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
for ax, (name, pred) in zip(axes, [('Vote OR', vote_or),
                                    ('Vote Majority', vote_maj),
                                    ('Vote AND', vote_and)]):
    cm = confusion_matrix(y_test, pred)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Purples', cbar=False,
                xticklabels=['Normal','Anomali'], yticklabels=['Normal','Anomali'], ax=ax)
    ax.set_xlabel("Prediksi"); ax.set_ylabel("Ground Truth")
    ax.set_title(f"Confusion Matrix - {name}", fontweight='bold')
plt.tight_layout(); plt.show()
```

```python
voting_metrics = []
for name, preds in [('Vote OR', vote_or), ('Vote Majority', vote_maj), ('Vote AND', vote_and)]:
    pre = precision_score(y_test, preds, zero_division=0)
    rec = recall_score(y_test, preds, zero_division=0)
    f1  = f1_score(y_test, preds, zero_division=0)
    voting_metrics.append({'Strategi': name, 'Precision': pre, 'Recall': rec, 'F1-Score': f1})

voting_metrics_df = pd.DataFrame(voting_metrics)
print("--- Ringkasan Metrik Antar Strategi Voting pada Test Set ---")
print(voting_metrics_df.to_string(index=False))

fig, ax = plt.subplots(figsize=(10, 5))
x = np.arange(len(voting_metrics_df))
width = 0.25

bars_p = ax.bar(x - width, voting_metrics_df['Precision'].values, width, label='Precision', color='#1f77b4', edgecolor='black')
bars_r = ax.bar(x,         voting_metrics_df['Recall'].values,    width, label='Recall',    color='#ff7f0e', edgecolor='black')
bars_f = ax.bar(x + width, voting_metrics_df['F1-Score'].values,  width, label='F1-Score',  color='#2ca02c', edgecolor='black')

ax.set_xlabel('Strategi Voting'); ax.set_ylabel('Skor')
ax.set_title('Perbandingan Precision, Recall, dan F1-Score Antar Strategi Voting (Test Set)', fontweight='bold')
ax.set_xticks(x); ax.set_xticklabels(voting_metrics_df['Strategi'].values)
ax.set_ylim(0, 1.05); ax.legend(loc='lower right'); ax.grid(axis='y', linestyle='--', alpha=0.4)

for bars in [bars_p, bars_r, bars_f]:
    for bar in bars:
        ax.annotate(f'{bar.get_height():.3f}',
                    xy=(bar.get_x() + bar.get_width() / 2, bar.get_height()),
                    xytext=(0, 3), textcoords='offset points',
                    ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.show()
```

### Step 14.C - Interpretasi Ensemble Voting

Tabel dan grafik di atas memberikan beberapa insight penting:

**Trade-off precision vs recall antar strategi voting:**
- **Vote OR** cenderung memiliki recall tertinggi (paling sensitif), tetapi dapat mengorbankan precision karena cukup satu model salah-deteksi untuk menghasilkan False Positive.
- **Vote AND** cenderung memiliki precision tertinggi (paling konservatif), tetapi dapat mengorbankan recall karena ketiga model harus setuju.
- **Vote Majority** berada di tengah dan biasanya menghasilkan F1-Score yang paling seimbang.

**Kapan ensemble unggul dari model tunggal?** Ensemble dapat mengungguli model tunggal bila dua syarat terpenuhi: (1) **diversitas error** — model-model salah pada hari yang berbeda, dan (2) **kekuatan sebanding** — tidak ada model yang terlalu dominan sehingga "menarik turun" yang lain.

**Rekomendasi praktis untuk monitoring kualitas air:**
- Untuk *early warning system* yang tidak boleh melewatkan pencemaran → **Vote OR**.
- Untuk sistem peringatan publik yang tidak boleh membunyikan alarm palsu → **Vote AND**.
- Untuk evaluasi kinerja berimbang → **Vote Majority** atau model tunggal terbaik dari Step 13.

## Step 15 - Perbandingan Waktu Training dan Inferensi

Mengukur waktu training (fit pada training set) dan inferensi (predict pada test set) untuk ketiga model sebagai bahan diskusi skalabilitas. Pengukuran memakai hyperparameter terbaik hasil tuning (Step 12) pada feature set pemenang.

```python
import time

feats   = winner_features
X_train = train_s[feats].values
X_test  = test_s[feats].values

timing_results = []
base_train_time = {}
base_inf_time   = {}

models_to_time = [
    ('LOF (Tuned)',     LocalOutlierFactor(novelty=True, **best_lof_params)),
    ('OCSVM (Tuned)',   OneClassSVM(kernel='rbf', **best_svm_params)),
    ('iForest (Tuned)', IsolationForest(random_state=RANDOM_STATE, **best_iso_params)),
]

for name, model in models_to_time:
    t0 = time.perf_counter()
    model.fit(X_train)
    t_train = time.perf_counter() - t0

    t0 = time.perf_counter()
    _ = model.predict(X_test)
    t_inf = time.perf_counter() - t0

    base_train_time[name] = t_train
    base_inf_time[name]   = t_inf
    timing_results.append({
        'Model': name,
        'Waktu Training (detik)':  t_train,
        'Waktu Inferensi (detik)': t_inf,
    })

total_train = sum(base_train_time.values())
total_inf   = sum(base_inf_time.values())

for strat in ['Ensemble Vote OR', 'Ensemble Vote Majority', 'Ensemble Vote AND']:
    timing_results.append({
        'Model': strat,
        'Waktu Training (detik)':  total_train,
        'Waktu Inferensi (detik)': total_inf,
    })

timing_df = pd.DataFrame(timing_results)
print(f"Jumlah titik pada test set: {len(X_test)}")
timing_df
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd', '#8c564b']

train_values = timing_df['Waktu Training (detik)'].values
axes[0].barh(timing_df['Model'], train_values, color=colors[:len(timing_df)], edgecolor='black')
axes[0].set_xlabel('Waktu (detik)'); axes[0].set_title('Waktu Training', fontweight='bold')
axes[0].grid(axis='x', linestyle='--', alpha=0.4)
max_train = train_values.max()
axes[0].set_xlim(0, max_train * 1.18)
for i, v in enumerate(train_values):
    axes[0].text(v + max_train * 0.01, i, f'{v:.4f} s', va='center', fontsize=9)

inf_values = timing_df['Waktu Inferensi (detik)'].values
axes[1].barh(timing_df['Model'], inf_values, color=colors[:len(timing_df)], edgecolor='black')
axes[1].set_xlabel('Waktu (detik)')
axes[1].set_title(f'Waktu Inferensi ({len(X_test)} titik)', fontweight='bold')
axes[1].grid(axis='x', linestyle='--', alpha=0.4)
max_inf = inf_values.max()
axes[1].set_xlim(0, max_inf * 1.18)
for i, v in enumerate(inf_values):
    axes[1].text(v + max_inf * 0.01, i, f'{v:.4f} s', va='center', fontsize=9)

plt.suptitle('Perbandingan Waktu Komputasi Antar Model dan Strategi Ensemble',
             fontsize=13, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```
